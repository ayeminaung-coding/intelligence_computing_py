## Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

#### Step 1: Create a Matrix
- Create matrix array "a2" 3x4 matrix with values [[12,13,14,15], [17, 18, 19, 20],[22, 23, 24, 25]]

```python
a2 = np.array([[12, 13, 14, 15],
               [17, 18, 19, 20],
               [22, 23, 24, 25]])
a2
```

#### Step 2: Slice Rows (using slicing : , for example a2[:3, :] → rows 0–2, all columns)
- Select the first 2 rows of matrix a2 
- 

```python
a2_first_rows = a2[:2, :]
a2_first_rows
```

#### Step 3: Slice Columns (using slicing : , for example a2[:3, :] → rows 0–2, all columns)
- Select the last 2 columns of matrix a2
- 

```python
a2_last_cols = a2[:, -2:]
a2_last_cols
```

#### Step 4: Reshape the Matrix
- Change the shape from (3 × 4) to (4 × 3)change reshape from 4 row 3 column
- 

```python
a2_reshaped = a2.reshape(4, 3)
a2_reshaped
```

#### Step 5: Column Selection
- Select column 2 and all row
- 

```python
col2 = a2_reshaped[:, 1:2]
col2
```

#### Step 6: Flatten the Matrix
- Change array matrix before to 1D array (flatten)

```python
flat_a2 = a2_reshaped.flatten()
flat_a2
```

#### Step 7: Find summation in array metrix "a2" using funtion sum()

```python
total_sum = np.sum(a2)
total_sum
```
